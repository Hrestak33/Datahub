```python
# renommer les colonnes pour les normaliser (minuscules, underscores, etc.)
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.columns
```

```python
df[df.columns]
```

```python
#convertir les colonnes binaires (Yes/No) en booleens (True/False)
df[df.columns] = df[df.columns].replace({'Yes': True, 'No': False})
```

```python
df.info()
```

```python
#Apercu de la nouvelle transformation dans le dataset true/false 
df.head()
```

```python
#7 filtrer les lignes ou la colonne 'Salary Estimate' est manquante 
salaire_manquant = df[df['salary_estimate'].isna()]
```

```python
#8 remplacer les valeurs manquantes dans la colonne 'YOE' par la moyenne
df['yoe'] = df['yoe'].fillna(df['yoe'].mean())
```

```python
# 9 Supprimer les lignes où 'Rating' est manquante
df.dropna(subset=['rating'], inplace=True)
```

```python
#10 Créer 'has_phd' (1 si True, sinon 0)
df['has_phd'] = df['phd'].astype(int)
```

```python
#diviser la colonne 'Salary Estimate' en deux colonnes 'salary_low' et 'salary_high' en
#utilisant uniquement str.split() et str.replace()
#On nettoie les symboles ($ et K)
clean_salary = df['salary_estimate'].str.replace('$', '').str.replace('K', '')
```

```python

df[['salary_low', 'salary_high']] = clean_salary.str.split('-', expand=True)
```

```python
#12 Convertir en valeurs numériques
df['salary_low'] = pd.to_numeric(df['salary_low'])
df['salary_high'] = pd.to_numeric(df['salary_high'])
```

```python
#13 

df['salary_mean'] = (df['salary_low'] + df['salary_high']) / 2
```

```python
# 14 Calculer l'écart de salaire
df['salary_gap'] = df['salary_high'] - df['salary_low']
```

```python
#15 Identifier les écarts > 50K

grands_ecarts = df[df['salary_gap'] > 50]
```
